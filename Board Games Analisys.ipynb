### Dataset Attributes

* **(ID) Board Game ID :** 

    - A unique identifier for each board game (Number).

* **(Name) Name:** 

    - The title of the board game (Text).

* **(Year_Published) Year Published:** 

    - The year the game was first published (Number).

* **(Min_Players) Minimum Number of Players:** 

    - The minimum number of players recommended to play the game (Number).

* **(Max_Players) Maximum Number of Players:** 

    - The maximum number of players recommended to play the game (Number).

* **(Playing_Time) Playing Time(Playing_Time):** 

    - The average duration required to play the game, usually in minutes (Number).

* **(Min_Age) Recommended Minimum Age:** 

    - The minimum age recommended for players (Number).

* **(Users_Rating)Number of User Ratings:** 

    - The total number of users who have rated the game (Number).

* **(Rating_Averege) Average Rating:** 

    - The average rating the game has received from users (Floating Point).

* **(BGG_Rank) BGG Rank:** 

    - The rank of the game on the BGG website as of the collection date (Number).

* **(Complexity_Avarege) Average Complexity:** 

    - A measure of the game's complexity based on user ratings, often reflecting the game's difficulty (Floating Point).

* **(Owned_Users) Number of BGG Registered Owners:** 

    - The number of BGG users who have registered ownership of the game (Number).

* **(Mechanics) Mechanics:** 

    - The various mechanics or systems used in the game (Text).

* **(Domains) Domains:** 

    - The categories or domains that the game belongs to (Text).

```python
# Calcular IQR para Play_Time                                               #########################################                        
Q1 = df['Play_Time'].quantile(0.25)                                         #                                       #
Q3 = df['Play_Time'].quantile(0.75)                                         #                                       #
IQR = Q3 - Q1                                                               #                                       #
                                                                            #   Técnica para filtrar os Outliners   #
# Definir limites                                                           #                  IQR                  #
lower_bound = Q1 - 1.5 * IQR                                                #                                       #
upper_bound = Q3 + 1.5 * IQR                                                #                                       # 
                                                                            #########################################
# Filtrar dados                                                                                 
df_filtered = df[(df['Play_Time'] >= lower_bound) & (df['Play_Time'] <= upper_bound)]
```

### Scarlletplot

```python
df.plot(kind = 'scatter',
        x = 'Play_Time',
        y = 'Rating_Average', 
        title = 'Com Outliners')
plt.show()
```

#### 2. Top 10 Average Rating Games:

### Is there a relationship between the Average Rating and the amount of rating the game receives?

#### 1. Observing The Correlations

```python
# Calcular correlação entre Rank e Quantidade de Ratings

correlation = df['Rating_Average'].corr(df['Users_Rating'])
print(f"Relationship Between The Average Rating And The Amount of Rating: {correlation:.3f}\n")

if abs(correlation) > 0.7:          #abs() pega o valor aboluto do número, como o módulo na matemática
    print("Strong correlation")
elif abs(correlation) > 0.3:
    print("Moderate correlation")
else:
    print("Weak or no correlation")
```
